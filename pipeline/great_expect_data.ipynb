# B. Instantiate Data Context

```python
# Create a data context

from great_expectations.data_context import FileDataContext

context = FileDataContext.create(project_root_dir='./')
```

# C. Connect to A `Datasource`

Connecting the Great Expectation model to data source

```python
# Give a name to a Datasource. This name must be unique between Datasources.
datasource_name = 'csv-data-bnpl-final'
datasource = context.sources.add_pandas(datasource_name)

# Give a name to a data asset
asset_name = 'clean_data_bnpl'
path_to_data = 'C:\Workspaces\Airflow_final_project\dags\clean_data.csv'
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)

# Build batch request
batch_request = asset.build_batch_request()
```

# D. Create an Expectation Suite

In this case, the `expectation_suite_name` with the name `expectation-bnpl-final-dataset` is used to store several expectations for the dataset. A ‘Validator’ is also used to generate expectations for the dataset.

```python
# Creat an expectation suite
expectation_suite_name = 'expectation-bnpl-final-dataset'
context.add_or_update_expectation_suite(expectation_suite_name)

# Create a validator using above expectation suite
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

# Check the validator
validator.head()
```

## D.1 - Expectations


**Expectations is basically what do we expect from the dataset.**

we expect our dataset have:
- Column 'employment_type' and 'customer_segment' has No Missing Value
- Column 'credit_score' must be between 300 until 850
- Column 'customer_segment' contain with Low Risk, Medium Risk, High Risk
- Column 'monthly_income' has No Missing Value
- Column 'monthly_income' must in type of integer or float (require)
- Column 'bnpl_installments' contain with 3, 6, 9, 12





Since it impossible for an employee not to have salary. So the 'salary' column must be ensured with no 'missing value'. 

```python
# Expectation 1a : Column `employment_type` can not contain missing values

validator.expect_column_values_to_not_be_null('employment_type')
```

```python
# Expectation 1b : Column `customer_segment` can not contain missing values

validator.expect_column_values_to_not_be_null('customer_segment')
```

```python
# Expectation 2 : Column `credit_score` must be between 0 and 5

validator.expect_column_values_to_be_between(
    column='credit_score', min_value=300, max_value=850
)
```

```python
# Expectation 3 : Column `customer_segment` must contain one of the following 3 things : Low Risk, Medium Risk, High Risk


validator.expect_column_values_to_be_in_set('customer_segment', ['Low Risk', 'Medium Risk', 'High Risk'])
```

```python
# Expectation 4 : Column `monthly_income` can not contain missing values

validator.expect_column_values_to_not_be_null('monthly_income')
```

```python
# Expectation 5 : Column `monthly_income` must in form of integer or float

validator.expect_column_values_to_be_in_type_list('monthly_income', ['integer','float'])
```

```python
# Expectation 6 : Column `bnpl_installments` must contain one of the following 3 things : 3, 6, 9, 12


validator.expect_column_values_to_be_in_set('bnpl_installments', [3, 6, 9, 12])
```

```python
# Save into Expectation Suite

validator.save_expectation_suite(discard_failed_expectations=False)
```

## D.2 - Checkpoint

```python
# Create a checkpoint

checkpoint_1 = context.add_or_update_checkpoint(
    name = 'checkpoint_m3',
    validator = validator,
)
```

```python
# Run a checkpoint

checkpoint_result = checkpoint_1.run()
```
